## Conditional Statements in Python

```python
marks = int(input("Enter The Marks : "))
if marks > 90 :
    print('Grade A')
    if marks > 90 :
        print('Highly eligible to apply for masters.')
elif marks <= 90 and marks >= 75 :
    print('Grade B')
elif marks < 75 and marks >= 60 :
    print('Grade C')
elif marks < 60 and marks >= 40 :
    print('Grade D')
    if marks < 60 and marks >= 40 :
        print("Improvement Needed")
else :
    print('Fail')
```

```python
age = int(input("Enter the Age"))
if age >= 18 :
    print('Candidate is eligible to vote.')
else :
    print('Candidate is not eligible to vote.')
```

```python
balance = 80000
transaction = int(input("Enter the amount to be withdrawn : "))
if transaction > balance or transaction == 0 :
    print('Invalid Transaction.')
    if transaction > balance :
        print('You have insufficient balance to withdraw money.')
else :
    print('Money Withdrawn and Remaining balance is :',balance - transaction)
```

```python
v1 = int(input("Enter 1st value : "))
v2 = int(input("Enter 2nd value : "))
operator = str(input("Enter the Operator : "))
if operator == '+':
    print('Addition Operation')
    if operator == '+':
        print(v1 + v2)
elif operator == '-':
    print('Substraction Operation')
    if operator == '-':
        print(v1 - v2)
elif operator == '*':
    print('Multiplication Operation')
    if operator == '*':
        print(v1 * v2)
elif operator == '/':
    print('Division Operation')
    if operator == '/':
        print(v1 / v2)
elif operator == '//':
    print('Int Division Operation')
    if operator == '//':
        print(v1 // v2)
elif operator == '%':
    print('Modulus Operation')
    if operator == '%':
        print(v1 % v2)
else:
    print('Invalid Operation')
```

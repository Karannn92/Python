## Looping in Python : 

1. For Loop

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,16):
    print(i,end = '  ')
```

```python
for i in range(3,10,2):
    print(i)
```

```python
for i in range(5,50,5):
    print(i,end = ' ')
```

```python
for i in range(100,1,-20):
    print(i,end = ' ')
```

```python
start = int(input("Enter the Value : "))
stop = int(input("Enter the Value : "))
step = int(input("Enter the Value : "))
for i in range(start,stop,step):
    print(i,end = ' ')
```

```python
d = {}
for i in range(3):
    k = input("Enter the Key : ")
    v = input("Enter the Value : ")
    d[k] = v
print(d)
print(type(d))
```

```python
n = 3 
for i in range(1,11):
    print(n, 'x', i, '=', n*i)
```

```python
# f- Formatted String
n = 3 
for i in range(1,11):
    print(f'{n} x {i} = {n*i}')
```

```python
pwd = 'user123'
for i in range(3):
    p = input("Enter the Password : ")
    if p != pwd:
        print("Incorrect Password!")
    else:
        print("Access Granted!")
        break
```

Break, Continue and Else Statements

```python
# Break Statement
for i in range(1,10):
    if i == 7:
        break
    print(i)
```

```python
# Else Statement
for i in range(1,16,4):
    if i == 7:
        break
    print(i)
else:
    print('Loop Ended')
```

```python
# Continue Statement
for i in range(1,10):
    if i == 8:
        continue
    print(i)
```

```python
for i in range(1,6):
    print("*" * i)
```

```python
n = 8
for i in range(1,n+1):
    print(" " * (n-i) + "*" * (2*i-1))
```

```python
n = 8
for i in range(1,n+1):
    print(" " * (n-i) + "*" * (2*i-1))
for j in range(n,0,-1):
    print(" " * (n-j) + "*" * (2*j-1))
```

```python
txt = 'madam'
ui = input("Enter the Text : ")
if ui[::-1] == txt:
    print('Is a palindrome text')
else:
    print('Is a non-palindrome text')
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
for l1 in list1:
    for l2 in list2:
        print(l1 + l2,end=" ")
    print()
```

```python
for i in range(1,10):
    for j in range(1,11):
        print(i*j,end = '\t')
    print()
```

```python
# factorial
n = 8
f = 1
for i in range(1,n+1):
    f *= i
    print(f)
```

```python
# fibonacci series
a = 0
b = 1
n = 10
for i in range(n):
    print(a, end=" ")
    a,b = b,a+b
```

```python
pw = 'user32'
ui = input("Enter the pw : ")
while pw != ui:
    print("Try Again")
    ui = input("Re-enter the pw : ")
print("Access Granted!")
```

```python
i = 1
while i <= 5:
    i += 1
    print(i)
```

```python
i = 7
while i >= 5:
    i -= 1
    print(i)
```

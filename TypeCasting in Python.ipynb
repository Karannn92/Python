## TypeCasting

- It means converting one data type into another data type.
- Eg : str -> Int, Int -> Float, str -> Float, Set -> List, etc.

1) String TypeCasting

```python
#str to Int
s1 = '32'
print(s1)
print(type(s1))
i1 = int(s1)
print(i1)
print(type(i1))
```

```python
s1
```

```python
i1
```

```python
# str to Float
s2 = '24.6'
print(s2)
print(type(s2))
f2 = float(s2)
print(f2)
print(type(f2))
```

```python
# str to Complex
s3 = '5+9j'
print(s3)
print(type(s3))
c3 = complex(s3)
print(c3)
print(type(c3))
```

```python
# str to Boolean
s4 = 'True'
print(s4)
print(type(s4))
b4 = bool(s4)
print(b4)
print(type(b4))
```

```python
s4 = ''
print(s4)
print(type(s4))
b4 = bool(s4)
print(b4)
print(type(b4))
```

```python
# str to List
s5 = 'Hello'
print(s5)
print(type(s5))
l5 = list(s5.split(","))
print(l5)
print(type(l5))
```

```python
# str to Tuple
s6 = 'Python'
print(s6)
print(type(s6))
t6 = tuple(s6.split(","))
print(t6)
print(type(t6))
```

```python
# str to Set
s7 = '11'
print(s7)
print(type(s7))
st7 = set(s7.split(","))
print(st7)
print(type(st7))
```

```python
# str to Frozenset
s8 = '23'
print(s8)
print(type(s8))
fs8 = frozenset(s8)
print(fs8)
print(type(fs8))
```

```python
# str to Dictionary
s9 = 'DS'
print(s9)
print(type(s9))
d9 = dict.fromkeys(s9,'Name')
print(d9)
print(type(d9))
```

```python
d9.keys()
```

```python
d9.values()
```

```python
d9.items()
```

## Integer Typecasting

```python
# int to float
i = 23
print(i)
print(type(i))
f = float(i)
print(f)
print(type(f))
```

```python
# int to str
i1 = 23
print(i1)
print(type(i1))
s1 = str(i1)
print(s1)
print(type(s1))
```

```python
# int to complex
i2 = 5
print(i2)
print(type(i2))
c2 = complex(i2)
print(c2)
print(type(c2))
print("Real and Imagainary Values : ",c2.real,c2.imag,sep = ',')
```

```python
# int to bool
i3 = 1
print(i3)
print(type(i3))
b3 = bool(i3)
print(b3)
print(type(b3))
```

```python
# int to list
i4 = 32
print(i4)
print(type(i4))
l4 = [i4]
print(l4)
print(type(l4))
```

```python
# int to tuple
i5 = 23
print(i5)
print(type(i5))
t5 = (i5,)
print(t5)
print(type(t5))
```

```python
# int to set
i6 = 64
print(i6)
print(type(i6))
st6 = {i6}
print(st6)
print(type(st6))
```

```python
# int to frozenset
i7 = 42
print(i7)
print(type(i7))
fs7 = frozenset([i7])
print(fs7)
print(type(fs7))
```

```python
# int to dictionary
i9 = 2
print(i9)
print(type(i9))
d9 = {'key' : i9}
print(d9)
print(type(d9))
```

## List Typecasting

```python
# list to str
l1 = ['hello']
print(l1)
print(type(l1))
sr1 = str(l1)
print(sr1)
print(type(sr1))
sr1
```

```python
# list to int 
l2 = [1,2,3]
print(l2)
print(type(l2))
i2 = l2[2]
print(i2)
print(type(i2))
```

```python
# list to float
l3 = [1.2,2.4,3.2]
print(l3)
print(type(l3))
f3 = l3[1]
print(f3)
print(type(f3))
```

```python
# list to complex
l4 = [1,2+4j,4+2j,5+8j]
print(l4)
print(type(l4))
c4 = l4[3]
print(c4)
print(type(c4))
```

```python
# list to bool 
l5 = [True]
print(l5)
print(type(l5))
b5 = l5[0]
print(b5)
print(type(b5))
```

```python
# list to tuple
l6 = list(input("Enter : ").split(','))
print(l6)
print(type(l6))
t6 = tuple(l6)
print(t6)
print(type(t6))
```

```python
#list to set
l7 = list(input("Enter : ").split(' '))
print(l7)
print(type(l7))
s7 = set(l7)
print(s7)
print(type(s7))
```

```python
#list to frozenset
l8 = [False,True,1,0,5,4,0]
print(l8)
print(type(l8))
fs8 = frozenset(l8)
print(fs8)
print(type(fs8))
```

```python
# list to dictionary
l9 = ['Data Science','Data Analysis']
print(l9)
print(type(l9))
d9 = dict.fromkeys(l9,20)
print(d9)
print(type(d9))
```

## Tuple Typecasting

```python
# tuple to list
t1 = (12,13,14)
print(t1)
print(type(t1))
l1 = list(t1)
print(l1)
print(type(l1))
```

```python
l1.append(15)
t1 = tuple(l1)
t1
```

```python
# tuple to set
t2 = tuple(input("Enter : ").split(","))
print(t2)
print(type(t2))
st2 = set(t2)
print(st2)
print(type(st2))
```

```python
# tuple to frozenset
t3 = (1,2.4,5,8,8,4,True,False,0)
print(t3)
print(type(t3))
fs3 = frozenset(t3)
print(fs3)
print(type(fs3))
```

```python
# tuple to dictionary
t4 = (1,3,5,6)
print(t4)
print(type(t4))
d4 = dict.fromkeys(t4,'Number')
print(d4)
print(type(d4))
```

## Dictionary Typecasting

```python
# dict to list
d1 = {'name' : 'DS','duration': '10 months'}
print(d1)
print(type(d1))
l1 = list(d1)
print(l1)
print(type(l1))
```

```python
# dict to tuple
d2 = {'name' : 'Karan', 'age' : 20, 'salary' : 50000}
print(d2)
print(type(d2))
t2 = tuple(d2)
print(t2)
print(type(t2))
```

## Set Typecasting

```python
# set to list
s1 = {1,True,0,1,'true',23}
print(s1)
print(type(s1))
lst1 = list(s1)
print(lst1)
print(type(lst1))
```

```python
# set to tuple
st2 = set(input("Enter Values : ").split(","))
print(st2)
print(type(st2))
t2 = tuple(st2)
print(t2)
print(type(t2))
```

```python
# set to frozenset
st3 = set(input("Enter Values : ").split(","))
print(st3)
print(type(st3))
fs3 = frozenset(st3)
print(fs3)
print(type(fs3))
```

```python
# set to dict
d3 = dict.fromkeys(st3,'Number')
print(d3)
print(type(d3))
```

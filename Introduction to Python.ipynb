```python
f = 3.2
f1 = 5.6
f2 = 9.3
print(f+f1+f2)
```

```python
f3 = float(input("Enter the Value : "))
print(f3)
print(type(f3))
```

```python
a = int(input("Enter the Value : "))
b = float(input("Enter the Value : "))

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a * a)
print(b * b)
```

## Complex 

```python
c = 5 + 7j
print(c)
print(type(c))
```

```python
print("Real Number is : ",c.real)
print("Imaginary Number is : ",c.imag)
```

```python
c1 = 5+7j+2+3+9j
print(c1)
print(type(c1))
print(c1.real)
print(c1.imag)
```

```python
c2 = complex(input("Enter the Value : "))
c3 = complex(input("Enter the Value : "))
print(c2)
print(c3)
print(c2+c3)
```

```python
# Int Float and Complexx Combined Operation
i7 = int(input("Enter the Value : "))
i8 = float(input("Enter the Value : "))
i9 = complex(input("Enter the Value : "))
print(i7+i8+i9-i7/i8)
```

```python
st1 = {1, True, 0, False, 0}
print(st1)
```

```python
st
```

```python
st.add(3)
print(st)
```

```python
st.update([4,5])
print(st)
```

```python
st.pop()
```

```python
st.remove('hello')
print(st)
```

```python
st2 = {1, 2, 3}
st3 = {2, 3, 4, 5, 6}
```

```python
print(st2.union(st3))
```

```python
print(st2.intersection(st3))
```
